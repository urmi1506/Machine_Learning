```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
db = pd.read_csv("admission.csv")
```

```python
db.head()
```

```python
db.describe()
```

```python

fig, axes = plt.subplots(2, 2, figsize=(16,8))

axes[0,0].set_title("Distribution of GRE Score", fontsize=14)
axes[0,0].hist(db["GRE Score"], color="#FF9999", edgecolor="black", alpha=0.8)

axes[0,1].set_title("Distribution of TOEFL Score", fontsize=14)
axes[0,1].hist(db["TOEFL Score"], color="#66B3FF", edgecolor="black", alpha=0.8)

axes[1,0].set_title("Distribution of University Rating", fontsize=14)
axes[1,0].hist(db["University Rating"], color="#99FF99", edgecolor="black", alpha=0.8)

axes[1,1].set_title("Distribution of SOP", fontsize=14)
axes[1,1].hist(db["SOP"], color="#FFCC99", edgecolor="black", alpha=0.8)

plt.tight_layout()
plt.show()
```

```python
db.isnull()
```

```python
db.fillna(db.mean(), inplace = True)
```

```python
db.head()
```

```python
X = db[["GRE Score", "TOEFL Score", "University Rating", "SOP", "CGPA"]]
y = db["Research"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 42)
```

```python
lin_reg = LinearRegression()
```

```python
lin_reg.fit(X_train, y_train)
```

```python
y_pred = lin_reg.predict(X_test)
```

```python
print("Mean absolute error: ", metrics.mean_absolute_error(y_test, y_pred))
print("Mean squared error: ", metrics.mean_squared_error(y_test, y_pred))
print("R2 score: ", metrics.r2_score(y_test, y_pred))
```
