Data Visualization



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
db=pd.read_csv("heart.csv")
```

a) Compute and display summary statistics for each feature available in the dataset. (e.g. minimum value, maximum value, mean, range, standard deviation, variance and percentiles




```python
db.describe()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=db, x='age', kde=True,bins=20, color='teal')
plt.title('Distribution of Age', fontsize=16)
plt.xlabel('Age', fontsize=14)
plt.ylabel('Frequency', fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=db, x='chol', kde=True, bins=20, color='purple')
plt.title('Distribution of Cholesterol', fontsize=16)
plt.xlabel('Cholesterol', fontsize=14)
plt.ylabel('Frequency', fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='sex', data=db, palette="coolwarm")
plt.title("Sex Distribution (0 = Female, 1 = Male)", fontsize=14)
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(6, 6))
db['sex'].value_counts().plot.pie(autopct='%1.1f%%', colors=['lightblue', 'lightcoral'], labels=['Female', 'Male'])
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(6, 6))
db['target'].value_counts().plot.pie(autopct='%1.1f%%', colors=['green', 'red'], labels=['No Disease', 'Disease'])
plt.title("Heart Disease Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=db['age'], y=db['chol'], hue=db['target'], palette=['yellow', 'red'])
plt.title("Age vs Cholesterol (Colored by Heart Disease)")
plt.xlabel("Age")
plt.ylabel("Cholesterol Level")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x=db['age'], y=db['thalach'], hue=db['target'], palette=['blue', 'red'])
plt.title("Age vs Max Heart Rate (thalach)")
plt.xlabel("Age")
plt.ylabel("Max Heart Rate")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x=db['target'], y=db['chol'], palette="coolwarm")
plt.title("Cholesterol Levels by Heart Disease Presence")
plt.xlabel("Heart Disease (0 = No, 1 = Yes)")
plt.ylabel("Cholesterol Level")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x=db['target'], y=db['thalach'], palette="coolwarm")
plt.title("Max Heart Rate by Heart Disease Presence")
plt.xlabel("Heart Disease (0 = No, 1 = Yes)")
plt.ylabel("Max Heart Rate")
plt.show()
```
